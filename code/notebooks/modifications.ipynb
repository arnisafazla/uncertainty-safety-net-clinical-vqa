```python
import pandas as pd
import os

base_folder = "outputs/GMAI-MMBench/N_210-K_4-M_8"
```

# +nota: Add a new option in a random place

```python
import random, ast
def add_none_option_random(shuffled_options):
    # get a random number between 0 and len(shuffled_options)+1
    insert_position = random.randint(0, len(shuffled_options)+1)
    # get a list of values, in the correct order
    values = list(shuffled_options.values())
    # insert the new option at the random position
    values.insert(insert_position, "none of the options is correct")
    # create new keys
    new_keys = [chr(ord('A') + i) for i in range(len(values))]
    # create new shuffled_options dict
    new_shuffled_options = {new_keys[i]: values[i] for i in range(len(values))}
    return new_shuffled_options

def fix_shuffled_correct(row):
    new_row = row.copy()
    correct = row.original_options[row.original_correct_answer]
    for option in ["A", "B", "C", "D", "E", "F"]:
        if row.shuffled_options[option] == correct:
            new_row["shuffled_correct_answer"] = option
            return new_row
```

```python
df_original = pd.read_csv(f"{base_folder}/filtered_data.tsv", sep="\t")
df_original["shuffled_options"] = df_original["shuffled_options"].apply(lambda x: add_none_option_random(ast.literal_eval(x)))
df_original["original_options"] = df_original["original_options"].apply(lambda x: ast.literal_eval(x))
df_original = df_original.apply(fix_shuffled_correct, axis=1)
os.makedirs(f"{base_folder}-plus_none", exist_ok=True)
df_original.to_csv(f"{base_folder}-plus_none/filtered_data.tsv", sep="\t", index=False)
```

# -correct+note: replace correct option with NOTA

```python
def replace_correct_option(shuffled_options, shuffled_correct_answer):
    shuffled_options[shuffled_correct_answer] = "none of the options is correct"
    return shuffled_options
```

```python
df_original = pd.read_csv(f"{base_folder}/filtered_data.tsv", sep="\t")
df_original["shuffled_options"] = df_original.apply(lambda row: replace_correct_option(ast.literal_eval(row["shuffled_options"]), row["shuffled_correct_answer"]), axis=1)
os.makedirs(f"{base_folder}-plus_none", exist_ok=True)
df_original.to_csv(f"{base_folder}-corrcet_to_none/filtered_data.tsv", sep="\t", index=False)
```
